### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.

```python
k = int(input("K = "))
t3 = [(0,4,8,12),(7,12,18),(6,12),(10,12,14,16)]
c = 0
t31 = []

for t in t3:
    for i in t:
        if i%k == 0:
            c = c + 1
    if len(t) == c:
        t31.append(t)
    c = 0
print(t31)
```

### 04) WAP to create a list of tuples from given list having number and its cube in each tuple.

### 05) WAP to find tuples with all positive elements from the given list of tuples.

```python
t5 = [(0,4,8,12),(7,-12,18),(-6,-12),(10,12,14,16)]
c = 0
t51 = []

for t in t5:
    for i in t:
        if i >= 0:
            c = c + 1
    if len(t) == c:
        t51.append(t)
    c = 0
print(t51)
```

```python
name = "raj"
age = 19

print("Name: " + name + ", Age: " + str(age))        # Using Concatenation (+) : ( String only )

print("Name:" , name , ", Age:" ,age)                # Using Comma-separated Values : ( Automatically handles type and adds a space between arguments.)

print("Name: " + name + ", Age: " ,age)              # Using Concatenation and Comma-separated

print("Name: %s, Age: %d" % (name, age))             # Using % Formatting: (Requires specifying the type (%s for string, %d for int, %f for float).)

print("Name: {}, Age: {}".format(name, age))         # Using .format() Method: (Automatically converts types and uses placeholders ({}).)
print("Age: {1}, Name: {0}".format(name, age))       # With Positional Arguments:
print("Name: {n}, Age: {a}".format(n=name, a=age))   # With Named Arguments:

print(f"Name: {name} , Age: {age}")                  # Using f-Strings (Python 3.6+):( The most modern and readable method. Allows inline expressions)
print(f"Name: {name.upper()}, Age: {age + 5}")       # can use method too

print("Name: " + repr(name) + ", Age: " + repr(age))
```
