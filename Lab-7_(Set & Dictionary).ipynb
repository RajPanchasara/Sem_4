### 09) WAP to check whether two given strings are anagram or not using set.

```python
s91 = 'hello'
s92 = 'lolhe'

def anagram(s1, s2):
    l1 = []
    l2 = []
    for i in s1:
        l1.append(i)
    l1.sort()
    for i in s2:
        l2.append(i)
    l2.sort()
    if(l1==l2):
        print('Strings are Anagram.')
    else:
        print('Strings are not Anagram.')

anagram(s91, s92)
```

### 11) WAP to count number of vowels in given string using set.

```python
s = 'raj panchasara'
vowels = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
c = 0

for i in s:
    if(i in vowels):
        c += 1

print(c)
```

### 12) WAP to check if a given string is binary string or not.

```python
s12 = input('Enter number String : ')

def Binary_String(s):
    c = 0
    for i in s:
        if i in ('0', '1'):
            c += 1
    if c == len(s12):
        print('Binary String.')
    else:
        print('Not Binary String.')

Binary_String(s12)
```

### 13) WAP to sort dictionary by key or value.

```python
d = {1:'a', 2:'c', 3:'d', 7:'b'}

# Sorted by key
k = dict(sorted(d.items()))
print(k)

# Sorted by value
v = dict(sorted(d.items(), key=lambda item: item[1]))
print(v)
```

### 14) WAP to find the sum of all items (values) in a dictionary given by user. (Assume: values are numeric)

```python
d2 = {101: 1, 102: 2, 103: 3}
sum = 0
for i in d2.values():
    sum += i
print(sum)
```

### 15) WAP to handle missing keys in dictionaries.
#### Example : Given, dict1 = {'a': 5, 'c': 8, 'e': 2} 
#### if you look for key = 'd', the message given should be 'Key Not Found', otherwise print the value of 'd' in dict1.

```python
d3 = {'a':5, 'c':8, 'e':2}
c = input('Enter key :')

if c in d3.keys() :
    print('value of', c, ':', d3[c])
else :
    print('Key Not Found')
```
