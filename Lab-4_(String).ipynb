# String

### 01) WAP to check whether the given string is palindrome or not.

```python
a = input("Enter String : ")

b = a[::-1]

if(a==b):
    print("palindrome")
else :
    print("Not palindrome")
```

### 02) WAP to reverse the words in the given string.

```python
a = input("Enter String : ")
l1 = a.split(" ")
l2 = l1[::-1]
b = ' '.join(l2)

print('Reversed String : \b',b)
```

### 07) WAP to capitalize the first and last character of each word in a string.

```python
def capitalize(a):
    return ' '.join( word[0].upper() + word[1:-1].lower() + word[-1].upper() if len(word)>1 else word.upper()
                     for word in a.split())

a = "hello world from pytho n"
b = capitalize(a)
print(b)
```

### 09) Check if the password and confirm password is same or not. 
### In case of only case's mistake, show the error message.

```python
p1 = input('Enter Password :')
p2 = input('Enter confirm password :')

if(p1 == p2) :
    print('Passwords match')
elif(p1.lower() == p2.lower()) :
    print('Case mistake')
else :
    print('Passwords do not match')
```

### 12) : Rearrange the given string. First lowercase then uppercase alphabets.
### input : EHlsarwiwhtwMV
### output : lsarwiwhtwEHMV

```python
str1 = input('Enter the string : ')
lower = ''
upper = ''
for i in str1 :
    if i.islower() :
        lower += i
    else :
        upper += i

str1 = lower + upper
print(str1)
```

```python
name = "raj"
age = 19

print("Name: " + name + ", Age: " + str(age))        # Using Concatenation (+) : ( String only )

print("Name:" , name , ", Age:" ,age)                # Using Comma-separated Values : ( Automatically handles type and adds a space between arguments.)

print("Name: " + name + ", Age: " ,age)              # Using Concatenation and Comma-separated

print("Name: %s, Age: %d" % (name, age))             # Using % Formatting: (Requires specifying the type (%s for string, %d for int, %f for float).)

print("Name: {}, Age: {}".format(name, age))         # Using .format() Method: (Automatically converts types and uses placeholders ({}).)
print("Age: {1}, Name: {0}".format(name, age))       # With Positional Arguments:
print("Name: {n}, Age: {a}".format(n=name, a=age))   # With Named Arguments:

print(f"Name: {name} , Age: {age}")                  # Using f-Strings (Python 3.6+):( The most modern and readable method. Allows inline expressions)
print(f"Name: {name.upper()}, Age: {age + 5}")       # can use method too

print("Name: " + repr(name) + ", Age: " + repr(age))
```
