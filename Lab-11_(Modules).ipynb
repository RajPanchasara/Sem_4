### 09) WAP to Calculate number of days between two given dates.

```python
def calculate_days_between_dates(date1: str, date2: str, date_format: str = "%Y-%m-%d"):
    d1 = datetime.datetime.strptime(date1, date_format)
    d2 = datetime.datetime.strptime(date2, date_format)
    return abs((d2 - d1).days)

date1 = input("Enter the first date (YYYY-MM-DD) : ")
date2 = input("Enter the second date (YYYY-MM-DD): ")

days = calculate_days_between_dates(date1, date2)

print(f"Number of days between {date1} and {date2}: {days}")
```

### 10) WAP to Find the day of the week of a given date.(i.e. wether it is sunday/monday/tuesday/etc.)

```python
def find_day_of_week(date1: str, date_format: str = "%Y-%m-%d"):
    d1 = datetime.datetime.strptime(date1, date_format)
    return d1.strftime("%A")

date1 = input("Enter the first date (YYYY-MM-DD) : ")
day = find_day_of_week(date1)
print(f'{date1} is {day}')
```
