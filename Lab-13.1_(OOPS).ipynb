### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

r = float(input("Enter the radius of the circle: "))
c1 = Circle(r)

print(f"Area of the circle: {c1.area():.2f}")
print(f"Perimeter of the circle: {c1.perimeter():.2f}")
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance=0.0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount:.2f}. New balance: {self.balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn {amount:.2f}. Remaining balance: {self.balance:.2f}")
        elif amount > self.balance:
            print("Insufficient balance.")
        else:
            print("Withdrawal amount must be positive.")

    def check_balance(self):
        print(f"Account Balance: {self.balance:.2f}")

account = BankAccount("123456789", "John Doe", 500.0)

account.check_balance()

account.deposit(200)

account.withdraw(100)

account.check_balance()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory:
    def __init__(self):
        self.items = {}
        
    def add_item(self, item_name, price, quantity):
        if item_name in self.items:
            self.items[item_name]["quantity"] += quantity  # Update quantity if item exists
        else:
            self.items[item_name] = {"price": price, "quantity": quantity}
        print(f"Added {quantity} of {item_name} to inventory.")

    def remove_item(self, item_name, quantity):
        if item_name in self.items:
            if self.items[item_name]["quantity"] >= quantity:
                self.items[item_name]["quantity"] -= quantity
                print(f"Removed {quantity} of {item_name}.")
                if self.items[item_name]["quantity"] == 0:
                    del self.items[item_name]  # Remove item if quantity reaches zero
                    print(f"{item_name} is now out of stock and removed from inventory.")
            else:
                print(f"Insufficient stock to remove {quantity} of {item_name}.")
        else:
            print(f"{item_name} not found in inventory.")

    def update_item(self, item_name, price=None, quantity=None):
        if item_name in self.items:
            if price is not None:
                self.items[item_name]["price"] = price
            if quantity is not None:
                self.items[item_name]["quantity"] = quantity
            print(f"Updated {item_name} details.")
        else:
            print(f"{item_name} not found in inventory.")

    def display_inventory(self):
        print("\nCurrent Inventory:")
        if self.items:
            for item, details in self.items.items():
                print(f"{item}: Price = ${details['price']:.2f}, Quantity = {details['quantity']}")
        else:
            print("Inventory is empty.")

inventory = Inventory()

inventory.add_item("Laptop", 1200, 5)
inventory.add_item("Mouse", 25, 10)

inventory.display_inventory()

inventory.remove_item("Mouse", 3)

inventory.update_item("Laptop", price=1100)

inventory.display_inventory()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute

    def add(self, other):
        total_minutes = self.minute + other.minute
        extra_hours = total_minutes // 60
        minutes = total_minutes % 60

        total_hours = self.hour + other.hour + extra_hours
        return Time(total_hours, minutes)

    def __str__(self):
        return f"{self.hour} hours) and {self.minute} minute(s)"

time1 = Time(2, 45)
time2 = Time(1, 30)
result = time1.add(time2)
print("Time 1:", time1)
print("Time 2:", time2)
print("Added Time:", result)
```
