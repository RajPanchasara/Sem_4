# List

### 01) WAP to find sum of all the elements in a List.

```python
l1 = [1,2,3,4,5,6]
sum = 0
for i in l1:
    sum = sum + i
print(sum)
```

### 02) WAP to find largest element in a List.

```python
l2 = [1,2,3,4,5,6]
max = l2[0]

for i in l2 :
    if i > max :
        max = i
        
print(max)
```

### 03) WAP to find the length of a List.

```python
l3 = [1,2,3,4,5]
length = len(l3)
print(length)
```

### 04) WAP to interchange first and last elements in a list.

```python
l4 = [1,2,3,4,5]

temp = l4[0]
l4[0] = l4[-1]
l4[-1] = temp

print(l4)
```

### 05) WAP to split the List into two parts and append the first part to the end.

```python
l5 = [1,2,3,4,5]

list1 = [i for i in range( 1 , int(len(l5)/2)+1 ) ]
list2 = [i for i in range( int(len(l5)/2)+1 , len(l5)+1 ) ]

newList = list2
newList.extend(list1)
print(newList)
```

### 06) WAP to interchange the elements on two positions entered by a user.

```python
l6 = [1,2,3,4,5,6]

p1 = int(input('Enter position 1 : '))
p2 = int(input('Enter position 2 : '))

temp = l6[p1]
l6[p1] = l6[p2]
l6[p2] = temp

print(l6)
```

### 17) WAP to create a list of common elements from given two lists.

```python
l1 = [1,2,3,4,5,6]
l2 = [4,5,6,7,8,9]

commonList = []

for i in l1 :
    for j in l2 :
        if i == j :
            commonList.append(i)

print(commonList)
```
